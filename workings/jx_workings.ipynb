## Step 6a: load and split

## Skeleton: steps 6a–6c as plain code
Plain notebook code (no classes) for loading, splitting and the two preprocessing steps, to adapt into classes in the library. These cells are self-contained: they don't import `computing4dc_hw5`.

```python
from pathlib import Path

import pandas as pd
from sklearn.model_selection import train_test_split

# relative to this notebook's folder (notebooks/); inside the library, build it from __file__ instead
DATA_PATH = Path("../src/computing4dc_hw5/sample_diabetes_mellitus_data.csv")
TARGET = "diabetes_mellitus"
```

### 6a: load and split

```python
df = pd.read_csv(DATA_PATH, index_col=0)  # column 0 is an unnamed row number
df = df.set_index("encounter_id")

df_train, df_test = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=df[TARGET],  # keeps the same diabetes rate in train and test
)

df_train.shape, df_test.shape
```

### 6b: drop rows with NaN in age, gender, ethnicity

```python
DROP_NA_COLUMNS = ["age", "gender", "ethnicity"]

df_train = df_train.dropna(subset=DROP_NA_COLUMNS)
df_test = df_test.dropna(subset=DROP_NA_COLUMNS)

df_train.shape, df_test.shape
```

### 6c: fill NaN in height, weight with the column mean

```python
FILL_MEAN_COLUMNS = ["height", "weight"]

train_means = df_train[FILL_MEAN_COLUMNS].mean()  # learned from train only

# fillna with a Series fills each column with its own value, and leaves other columns alone
df_train = df_train.fillna(train_means)
df_test = df_test.fillna(train_means)  # test gets train's means too

train_means
```
